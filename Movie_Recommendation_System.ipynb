```python
import ast

def convert(obj):
    L = []
  
    if isinstance(obj, str):
        try:
            obj = ast.literal_eval(obj)
        except:
            return []   
    
    for i in obj:
        if isinstance(i, dict) and 'name' in i:
            L.append(i['name'])
        elif isinstance(i, str):
            L.append(i)  
    return L
```

```python
import ast
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
def precision_at_k(movie, new, movies, similarity, k=5):
    idx = new[new['title'] == movie].index[0]

    # Get recommendations
    rec = sorted(list(enumerate(similarity[idx])),
                 reverse=True, key=lambda x: x[1])[1:k+1]

    recommended_titles = [new.iloc[i[0]].title for i in rec]

    # Get actual genres (ground truth) from movies df
    actual_genres = set(movies[movies['title'] == movie].iloc[0]['genres'])

    hits = 0
    for title in recommended_titles:
        rec_genres = set(movies[movies['title'] == title].iloc[0]['genres'])
        if len(actual_genres.intersection(rec_genres)) > 0:
            hits += 1

    precision = hits / k
    return precision
```

```python
print("Precision@5 for Gandhi:", precision_at_k("Gandhi", new, movies, similarity))
```

```python
def recall_at_k(movie, new, movies, similarity, k=5):
    idx = new[new['title'] == movie].index[0]

    rec = sorted(list(enumerate(similarity[idx])),
                 reverse=True, key=lambda x: x[1])[1:k+1]

    recommended_titles = [new.iloc[i[0]].title for i in rec]

    actual_genres = set(movies[movies['title'] == movie].iloc[0]['genres'])

    # count unique genre hits
    genre_hits = set()

    for title in recommended_titles:
        rec_genres = set(movies[movies['title'] == title].iloc[0]['genres'])
        common = actual_genres.intersection(rec_genres)
        genre_hits.update(common)

    recall = len(genre_hits) / len(actual_genres)
    return min(recall, 1.0)
```

```python
def f1_at_k(movie, new, movies, similarity, k=5):
    p = precision_at_k(movie, new, movies, similarity, k)
    r = recall_at_k(movie, new, movies, similarity, k)
    
    if p + r == 0:
        return 0
    
    return 2 * (p * r) / (p + r)
```

```python
def average_precision_at_k(movie, new, movies, similarity, k=5):
    idx = new[new['title'] == movie].index[0]
    
    rec = sorted(list(enumerate(similarity[idx])),
                 reverse=True, key=lambda x: x[1])[1:k+1]
    
    recommended_titles = [new.iloc[i[0]].title for i in rec]
    actual_genres = set(movies[movies['title'] == movie].iloc[0]['genres'])

    score = 0.0
    hits = 0

    for i, title in enumerate(recommended_titles, start=1):
        rec_genres = set(movies[movies['title'] == title].iloc[0]['genres'])
        
        if len(actual_genres.intersection(rec_genres)) > 0:
            hits += 1
            score += hits / i
    
    return score / k


def MAP_at_k(movie_list, new, movies, similarity, k=5):
    total = 0
    for movie in movie_list:
        total += average_precision_at_k(movie, new, movies, similarity, k)
    return total / len(movie_list)
```

```python
import math

def ndcg_at_k(movie, new, movies, similarity, k=5):
    idx = new[new['title'] == movie].index[0]

    rec = sorted(list(enumerate(similarity[idx])),
                 reverse=True, key=lambda x: x[1])[1:k+1]

    recommended_titles = [new.iloc[i[0]].title for i in rec]
    actual_genres = set(movies[movies['title'] == movie].iloc[0]['genres'])

    dcg = 0.0
    for i, title in enumerate(recommended_titles, start=1):
        rec_genres = set(movies[movies['title'] == title].iloc[0]['genres'])

        relevance = 1 if len(actual_genres.intersection(rec_genres)) > 0 else 0
        
        dcg += relevance / math.log2(i + 1)

    # Ideal DCG (all correct at top ranks)
    ideal_relevance = [1] * k
    idcg = sum([rel / math.log2(i + 1) for i, rel in enumerate(ideal_relevance, start=1)])

    return dcg / idcg if idcg > 0 else 0
```

```python
print("Precision@5:", precision_at_k("Gandhi", new, movies, similarity))
print("Recall@5:", recall_at_k("Gandhi", new, movies, similarity))
print("F1@5:", f1_at_k("Gandhi", new, movies, similarity))
print("NDCG@5:", ndcg_at_k("Gandhi", new, movies, similarity))
```
